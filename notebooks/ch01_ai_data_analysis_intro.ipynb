### 데이터 분석에 필요한 도구 불러오기

이 코드는 본격적으로 데이터를 분석하기 전에  
**필요한 Python 도구들을 불러오고, 데이터 폴더의 위치를 지정하는 준비 단계**이다.

쉽게 말하면,

> "분석 시작하기 전에 필요한 도구와 데이터 위치를 준비하자!"

라는 의미이다.

---

#### 1. `Path` 불러오기

`from pathlib import Path`

파일이나 폴더의 **경로(위치)**를 쉽게 다루기 위한 기능이다.

예를 들어,

`../data/raw`

처럼 데이터가 저장된 폴더의 위치를 지정할 때 사용할 수 있다.

---

#### 2. pandas 불러오기

`import pandas as pd`

`pandas`는 **표 형태의 데이터를 다루는 도구**이다.

CSV 파일을 불러오거나,
필요한 데이터를 골라내고,
합계나 평균을 계산할 때 사용한다.

`pd`는 pandas를 짧게 부르기 위한 별명이다.

예:

`pd.read_csv()`

→ CSV 파일을 읽어오는 기능

---

#### 3. NumPy 불러오기

`import numpy as np`

`NumPy`는 **숫자 계산을 편리하게 해주는 도구**이다.

많은 숫자를 계산하거나 수학적인 처리가 필요할 때 사용한다.

`np`는 NumPy를 짧게 부르는 별명이다.

---

#### 4. Matplotlib 불러오기

`import matplotlib.pyplot as plt`

데이터를 **그래프나 차트로 보여주기 위한 도구**이다.

예를 들어 매출 데이터를 막대그래프로 표현할 때 사용할 수 있다.

`plt`라는 짧은 이름으로 사용한다.

---

#### 5. Seaborn 불러오기

`import seaborn as sns`

Seaborn도 **데이터를 그래프로 표현하는 도구**이다.

Matplotlib을 기반으로 하며,
그래프를 보기 좋게 표현할 때 많이 사용한다.

`sns`라는 짧은 이름으로 사용한다.

---

#### 6. 데이터 폴더 위치 지정

`DATA_DIR = Path('../data/raw')`

분석에 사용할 원본 데이터가 들어있는 폴더의 위치를
`DATA_DIR`이라는 변수에 저장한다.

여기서 `..`은

> **현재 폴더에서 한 단계 위로 이동**

이라는 뜻이다.

따라서 현재 Notebook이 `notebooks` 폴더에 있다면

`notebooks → 한 단계 위 → data → raw`

순서로 이동해서 데이터 폴더를 찾는다.

---

#### 7. 그래프 기본 스타일 설정

`sns.set_theme(style='whitegrid')`

Seaborn으로 만드는 그래프의 기본 디자인을 설정한다.

`whitegrid`는

> **흰색 배경에 격자선이 있는 그래프**

스타일이다.

데이터 값을 비교하기 쉽게 만들어준다.

---

#### 8. 데이터 폴더 위치 확인

`print(DATA_DIR)`

내가 설정한 데이터 폴더의 위치를 화면에 출력한다.

예를 들어

`../data/raw`

가 출력되면 Python이 현재 데이터 위치를 이렇게 기억하고 있다는 뜻이다.

---

### 전체 흐름

필요한 Python 도구 불러오기

데이터가 있는 폴더 위치 지정  
↓  
그래프 디자인 설정  
↓  
데이터 폴더 위치 출력해서 확인
